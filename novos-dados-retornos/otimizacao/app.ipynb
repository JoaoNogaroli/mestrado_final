```python
data = yf.download(ativos, period="15y", auto_adjust=True)['Close']
```

```python
data.loc['2017']['NATU3.SA']
```

```python
anos = list(range(2015,2027))
anos
```

```python
dfs = {}
for ano in anos:
    dfs[ano] = data.loc[str(ano)]
    df_teste = dfs[ano].fillna(0)
    ativos_totais = (df_teste == 0).mean()
    ativos_manter = ativos_totais[ativos_totais<0.5].index
    dfs[ano] = df_teste[ativos_manter]
    dfs[ano] = (round(dfs[ano],2)/round(dfs[ano].shift(1),2))[1:]
    tamanho = len(dfs[ano].columns)

    termofinal = f'{str(ano)}-03-31'
    dfs[ano] = dfs[ano].loc[:termofinal]
    print(f"{ano} -> {tamanho}, Len: {len(dfs[ano])}, 1º Periodo: {dfs[ano].index[0]} | 2º Período {dfs[ano].index[-1]}")

    dfs[ano].to_csv(f'df_ativos_{ano}.csv')
```
