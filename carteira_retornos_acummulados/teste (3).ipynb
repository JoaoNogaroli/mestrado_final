```python
import numpy as np
import os
import pandas as pd
```

```python
dfs = {}
dfs[2015] = pd.read_csv("./sample_data/df_ativos_2015.csv").set_index('date')
dfs[2016] = pd.read_csv("./sample_data/df_ativos_2016.csv").set_index('date')

```

```python
print(dfs[2015])
print(dfs[2016])
```

```python
np.random.choice(dfs[2015].columns.tolist(),10)
```

```python
ativos_2015 = np.random.choice(dfs[2015].columns.tolist(),10,replace=False)
ativos_2016 = np.random.choice(dfs[2016].columns.tolist(),10,replace=False)
#
```

```python
pesos_2015 = np.random.rand(10)
pesos_2015 = pesos_2015/pesos_2015.sum()
pesos_2016 = np.random.rand(10)
pesos_2016 = pesos_2016/pesos_2016.sum()
```

```python
df = dfs[2015][ativos_2015]*pesos_2015
df_retorno_2015 = df.sum(axis=1)
df_retorno_2015 = (df_retorno_2015+1).cumprod()

print(df_retorno_2015)

df2 = dfs[2016][ativos_2015]*pesos_2016
df_retorno_2016= df2.sum(axis=1)
df_retorno_2016 = (df_retorno_2016+1).cumprod()
df_retorno_2016 = df_retorno_2016*df_retorno_2015.iloc[-1]
print(df_retorno_2016)
```

```python
final = pd.concat([df_retorno_2015,df_retorno_2016])
final*100
```
