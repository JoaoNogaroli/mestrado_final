```python
import pandas as pd
import os
```

```python
an = [f'20{i}{j}' for i in range(1,3) for j in range(10)]
anos = an[5:-4]
print(anos)
```

```python
for ano in anos:
    dc = {}
    for pasta in os.listdir('planilhas'):
        df = pd.read_excel('planilhas/' + pasta).set_index('Data')
        linha = df.loc[f'{ano}-12-31']

        for i in range(20):                 # <- o loop que faltava
            ticker = linha.iloc[i*3]
            roic = linha.iloc[i*3 + 1]
            ey = 1 / linha.iloc[i*3 + 2]
            dc[ticker] = {'ROIC': roic, 'EY': ey}

    df_final = pd.DataFrame(dc).T.dropna()  # <- só no fim, depois de todas as planilhas
    # df_final['magic_formula'] = df_final['ROIC'].rank(ascending=False) + df_final['EY'].rank(ascending=False)
    df_final['magic_formula'] = (0.50* df_final['ROIC']) + (0.50*df_final['EY'])
    df_final['magic_formula']= (df_final['magic_formula']-df_final['magic_formula'].min())/(df_final['magic_formula'].max()-df_final['magic_formula'].min())
    df_final.sort_values('magic_formula').to_excel(f'resultados_provisorio/{ano}.xlsx')
```
